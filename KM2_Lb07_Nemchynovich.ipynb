### Проектирование класса BinaryTree

Базовое определение класса `BinaryTree` содержит метод инициализации `__init__(self)`, метод строкового представления `__repr__(self)`, метод вставки элемента в дерево `insert(self, value)`.

Класс `BinaryTree` является основным. Он определяет шаблон для создания объекта, представляющего бинарное дерево поиска

* *Метод инициализации* экземпляра класса `__init__` вызывается без аргументов и инициализирует **единственный атрибут** экземпляра класса `root` экземпляром класса `EmptyNode`

Назначение атрибута `root` -- это хранение ссылки на корень дерева.

* *Метод строкового представления* экземпляра класса `__repr__` возвращает строковое представление для корня дерева `root`

* Метод `insert(value)` является методом вставки в дерево НОВОГО элемента. Новый элемент является листом со значением `value`.

Метод вставки элемента в дерево реализуется через метод вставки элемента в корневую вершину дерева `root`. При этом атрибут `root` переопределяется ссылкой на новый объект для представления корня

```python
class BinaryTree:
    def __init__(self):         # метод инициализации -- инициализирует экземпляр класса: метод обязан передать значения аргументов вызова атрибутам объекта.
        self.root = EmptyNode() # атрибут root, self - cтановится ссылкой на объект. Пока что у нас нет главной вершины, поэтому корень дерева -- экземпляр класса EmptyNode()
        
    def __repr__(self):         # метод строкового представления экземпляра класса
        return repr(self.root)  # возвращает строковое представление для корня дерева root (как он это делает -- определим позже)
        
    def insert(self, value):    # метод вставки элемента в дерево
        self.root = self.root.insert(value)   # переопределяем атрибут экземпляра класса root ссылкой на новый объект, происходит вставка в корневую вершину дерева root (insert определим позже)
```

При этом атрибут `root` `переопределяется ссылкой` на новый объект для представления корня

### Проектирование класса BinaryNode

Класс `BinaryNode` определяет **шаблон** для создания объекта, представляющего элемент бинарного дерева поиска

Определение класса `BinaryNode` содержит ТРИ метода: `__init__`, `__repr__` и `insert`.

* *Метод инициализации* экземпляра класса вызывается с тремя аргументами `left`, `value` и `right` для инициализации одноименных атрибутов экземпляра класса `left`, `value` и `right` соответствующими значениями

```python
def __init__(self, left, value, right): # метод инициализации -- обязан передать значения аргументов вызова атрибутам объекта
        self.left = left                # self -- ссылка на объект (в будущем вместо него можно будет условно подставлять имя экземпляра класса), атрибут left -- хранит ссылку на корень левого поддерева 
        self.value = value              # атрибут value -- хранит значение вершины
        self.right = right              # атрибут right -- хранит ссылку на корень правого поддерева 
```

Назначение атрибута `left` -- это хранение ссылки на корень левого поддерева. Назначение атрибута `right` -- это хранение ссылки на корень правого поддерева. Назначение атрибута `value` -- это хранение значения вершины.

* *Метод строкового представления экземпляра класса* `__repr__` возвращает кортеж, состоящий из строкового представления корня левого поддерева `left`, из строкового представления значения вершины `value` и из строкового представления корня правого поддерева `right`

```python
def __repr__(self):   # метод строкового представления экземпляра класса
    return "(%s, %s, %s)" % (repr(self.left), repr(self.value), repr(self.right)) # возвращает кортеж из строкового представления корня левой, правой вершин и значение вершины value
```

Элементы списка `source_data` будут добавляться в бинарное дерево поиска последовательно: сначала добавим значение `5`, потом значение `1` и т.д.

Сначала в дерево добавляется значение `source_data[0]` с помощью метода `insert` экземпляра класса `tree`. Метод `insert` класса `BinaryTree` вызывает метод `insert` для объекта, на который указывает атрибут `self.root` для корня. На текущем шаге корень представлен экземпляром класса `EmptyNode`. В результате создается экземпляр класса `BinaryNode`, который замещает пустую вершину в корне дерева.

Экземпляр класса `BinaryTree` не содержит поддеревьев, поэтому выводится в виде `(*,5,*)`

Далее в дерево добавляется значение `source_data[1]` с помощью метода `insert` экземпляра класса `tree`. Метод `insert` класса `BinaryTree` вызывает метод `insert` для корня. **На текущем шаге корень не пустой**, он представлен экземпляром класса `BinaryNode`.

$\color{red}{Алгоритм}$ метода `insert(self, value)` класса `BinaryNode` следующий:

* добавляемое значение `value` сравнивается со значением корня `self.value`;
* если `value<self.value`, то вызывается метод `insert(value)` для левого поддерева;
* если `value>=self.value`, то вызывается метод `insert(value)` для правого поддерева.

```python
class BinaryNode:
    def __init__(self, left, value, right): # метод инициализации -- обязан передать значения аргументов вызова атрибутам объекта
        self.left = left                    # self -- ссылка на объект (в будущем вместо него модно будет условно подставлять имя экземпляра класса), атрибут left -- хранит ссылку на корень левого поддерева 
        self.value = value                  # атрибут value -- храненит значение вершины
        self.right = right                  # атрибут right -- хранит ссылку на корень левого поддерева 
    def __repr__(self):                     # метод строкового представления экземпляра класса
        return "(%s, %s, %s)" % (repr(self.left), repr(self.value), repr(self.right)) # возвращает кортеж из строкового представления корня левой, правой вершин и значение вершины value
    def insert(self, value):                # метод вставки новой вершины
        if value < self.value:              # если найденное значение меньше значения, на которое в данный момент ссылается корень
            self.left = self.left.insert(value)  # вызывается метод insert(value) для левого поддерева
        else:
            self.right = self.right.insert(value)  # иначе вызывается метод insert(value) для правого поддерева
        return self
```

## Задание 7.4 Перегрузка операции принадлежности in

$\color{red}{Переопределим}$ классы `BinaryTree`, `BinaryNode`, `EmptyNode` за счет определения нового метода `__contains__(self, value)` для перегрузки операции принадлежности `in`.

Операция `in`, вызванная для экземпляра `tree` класса `BinaryTree` в виде $$ value`in`tree $$ будет возвращать `True`, если значение `value` найдено в дереве, и `False`, если значение `value` не найдено в дереве.

Встроенная операция принадлежности `in` вызывает метод `__contains__`, поэтому этот метод необходимо определить для трех классов `BinaryTree`, `EmptyNode`, `BinaryNode`.

Метод `__contains__` класса `BinaryTree` возвращает результат вызова операции `in` для корневой вершины

```python
class BinaryTree:
    def __init__(self):         # метод инициализации -- инициализирует экземпляр класса: метод обязан передать значения аргументов вызова атрибутам объекта.
        self.root = EmptyNode() # атрибут root, self - cтановится ссылкой на объект. Пока что у нас нет главной вершины, поэтому корень дерева -- EmptyNode()
        
    def __repr__(self):         # метод строкового представления экземпляра класса
        return repr(self.root)  # возвращает строковое представление для корня дерева root (как он это делает -- определим позже)
        
    def insert(self, value):    # метод вставки элемента в дерево
        self.root = self.root.insert(value)   # переопределяем атрибут экземпляра класса root, происходит вставка в корневую вершину дерева root (insert определим позже)
        
    def __contains__(self, value):  # метод принадлежности
        return value in self.root   # вернет True, если значение value найдено в дереве, иначе False
```

Метод `__contains__` класса `EmptyNode` возвращает `False`

```python
class EmptyNode:        
    def __repr__(self):   # метод строкового представления экземпляра класса
        return '*'        # возвращает *, если экземпляр класса EmptyNode
    
    def insert(self, value):                   # метод вставки новой вершины
        return BinaryNode(self, value, self)   # возвращает новый экземпляр класса BinaryNode со значением value, левое и правое поддерево которого определяются пустой вершиной
    
    def __contains__(self, value): # метод принадлежности
        return False               # вернет False, так как в экземпляре класса EmptyNode нет элементов (он пустой)
```

$\color{red}{Алгоритм}$ метода `__contains__(self, value)` класса `BinaryNode` очень похож на алгоритма метода `insert`:

* проверяемое значение `value` сравнивается со значением корневой вершины `self.value`;
* если `value==self.value`, то возвращается `True`;
* если `value<self.value`, то возвращается результат вызова операции `in` для корня левого поддерева;
* если `value>self.value`, возвращается результат вызова операции `in` для корня правого поддерева.

```python
class BinaryNode:
    def __init__(self, left, value, right): # метод инициализации -- обязан передать значения аргументов вызова атрибутам объекта
        self.left = left                    # self -- ссылка на объект (в будущем вместо него модно будет условно подставлять имя экземпляра класса), атрибут left -- хранит ссылку на корень левого поддерева 
        self.value = value                  # атрибут value -- храненит значение вершины
        self.right = right                  # атрибут right -- хранит ссылку на корень левого поддерева 
        
    def __repr__(self):                     # метод строкового представления экземпляра класса
        return "(%s, %s, %s)" % (repr(self.left), repr(self.value), repr(self.right)) # возвращает кортеж из строкового представления корня левой, правой вершин и значение вершины value
    
    def insert(self, value):                # метод вставки новой вершины
        if value < self.value:
            self.left = self.left.insert(value)
        else:
            self.right = self.right.insert(value)
        return self
    
    def __contains__(self, value):  # метод принадлежности
        if value < self.value:
            return value in self.left
        elif value > self.value:
            return value in self.right
        else:
            return True
```

## Задание 7.5 Перегрузка встроенной функции len

$\color{red}{Переопределим}$ классы `BinaryTree`, `BinaryNode`, `EmptyNode` за счет определения нового метода `__len__(self)` для перегрузки встроенной функции `len`, которая возвращает количество вершин в бинарном дереве поиска.

Встроенная функция `len`, вызванная от экземпляра класса `BinaryTree`, будет возвращать количество элементов бинарного дерева поиска. Функция `len` вызывает метод `__len__`, поэтому этот метод необходимо определить для трех классов `BinaryTree`, `EmptyNode`, `BinaryNode`.

Метод `__len__` класса `BinaryTree` вызывает функцию `len` для корневой вершины

```python
class BinaryTree:
    def __init__(self):         # метод инициализации -- инициализирует экземпляр класса: метод обязан передать значения аргументов вызова атрибутам объекта.
        self.root = EmptyNode() # атрибут root, self - cтановится ссылкой на объект. Пока что у нас нет главной вершины, поэтому корень дерева -- EmptyNode()
        
    def __repr__(self):         # метод строкового представления экземпляра класса
        return repr(self.root)  # возвращает строковое представление для корня дерева root (как он это делает -- определим позже)
        
    def insert(self, value):    # метод вставки элемента в дерево
        self.root = self.root.insert(value)   # переопределяем атрибут экземпляра класса root, происходит вставка в корневую вершину дерева root (insert определим позже)
        
    def __contains__(self, value):
        return value in self.root
    
    def __len__(self):
        return len(self.root)
```

Метод `__len__` класса `EmptyNode` возвращает `0`

```python
class EmptyNode:        
    def __repr__(self):   # метод строкового представления экземпляра класса
        return '*'        # возвращает *, если экземпляр класса EmptyNode
    
    def insert(self, value):                   # метод вставки новой вершины
        return BinaryNode(self, value, self)   # возвращает новый экземпляр класса BinaryNode со значением value, левое и правое поддерево которого определяются пустой вершиной
    
    def __contains__(self, value):
        return False
    
    def __len__(self):
        return 0
```

Метод `__len__` класса `BinaryNode` возвращает значение атрибута `count` класса `BinaryNode`, который используется как счетчик созданных экземпляров класса `BinaryNode`

```python
class BinaryNode:
    
    def __init__(self, left, value, right): # метод инициализации -- обязан передать значения аргументов вызова атрибутам объекта
        self.left = left                    # self -- ссылка на объект (в будущем вместо него модно будет условно подставлять имя экземпляра класса), атрибут left -- хранит ссылку на корень левого поддерева 
        self.value = value                  # атрибут value -- храненит значение вершины
        self.right = right                  # атрибут right -- хранит ссылку на корень левого поддерева 
        self.count = 1
    
    def __repr__(self):                     # метод строкового представления экземпляра класса
        return "(%s, %s, %s)" % (repr(self.left), repr(self.value), repr(self.right)) # возвращает кортеж из строкового представления корня левой, правой вершин и значение вершины value
    
    def insert(self, value):                # метод вставки новой вершины
        self.count+=1                       # к атрибуту count добавляется 1 при вставке нового элемента в дерево
        if value < self.value:
            self.left = self.left.insert(value)
        else:
            self.right = self.right.insert(value)
        return self
    
    def __contains__(self, value):
        if value < self.value:
            return value in self.left
        elif value > self.value:
            return value in self.right
        else:
            return True
    
    def __len__(self):
        return self.count
```
